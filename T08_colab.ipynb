# T08 · ¿Este jugador profesional se quema en el split?
Versión final corregida después de la corrida 1. Las reglas se fijan con entrenamiento (casos 1–14); prueba (15–20) se usa solo para medir.

```python
def predecir_burnout(dias_libres, horas_sueno, partidas_oficiales, horas_practica):
    if partidas_oficiales >= 20:
        return 1, 'R1: carga competitiva alta (20 o más partidas oficiales)'
    if dias_libres <= 2 and horas_practica >= 11:
        return 1, 'R2: muy pocos días libres + práctica alta'
    return 0, 'Ninguna regla de riesgo se activó'

print('Sistema T08 final cargado.')
```

```python
# EJECUTAR ESTA CELDA EN LA MESA
dias_libres = float(input('Días libres en el mes: '))
horas_sueno = float(input('Horas de sueño promedio: '))
partidas_oficiales = float(input('Partidas oficiales en el mes: '))
horas_practica = float(input('Horas de práctica por día: '))

pred, motivo = predecir_burnout(dias_libres, horas_sueno, partidas_oficiales, horas_practica)
print('\nSALIDA:', 'SÍ' if pred == 1 else 'NO')
print('Motivo:', motivo)
```
